```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Aggregate revenue by sale date or month
df['sale_date'] = pd.to_datetime(df['sale_date'])
monthly_sales = df.groupby(df['sale_date'].dt.to_period('M'))['revenue_usd'].sum()

# Plot line chart
plt.figure(figsize=(8, 4))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o', color='b')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month", fontsize=10)
plt.ylabel("Total Revenue (USD)")
plt.grid(True)

# Reduce clutter on X-axis
plt.xticks(rotation=45, fontsize=8)  # Rotate labels and shrink font size
plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(nbins=6))  # Limit visible ticks to max 6

plt.tight_layout()  # Adjust padding so rotated labels fit cleanly
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

df['sale_date'] = pd.to_datetime(df['sale_date'])
df['month'] = df['sale_date'].dt.to_period('M').astype(str)

# Calculate monthly revenue and total units sold
monthly_metrics = df.groupby('month')[['units_sold', 'revenue_usd']].sum()

fig, ax1 = plt.subplots(figsize=(8, 4))

ax1.set_xlabel('Month', fontsize=10)
ax1.set_ylabel('Revenue (USD)', color='tab:blue')
ax1.plot(monthly_metrics.index, monthly_metrics['revenue_usd'], color='tab:blue', label='Revenue')

ax2 = ax1.twinx()  
ax2.set_ylabel('Units Sold', color='tab:orange')
ax2.plot(monthly_metrics.index, monthly_metrics['units_sold'], color='tab:orange', label='Units')

# Reduce clutter on X-axis
ax1.tick_params(axis='x', rotation=45, labelsize=8)  # Rotate and reduce font size
ax1.xaxis.set_major_locator(ticker.MaxNLocator(nbins=6))  # Show at most 6 ticks

plt.title("Monthly Revenue vs Units Sold Trend")
plt.tight_layout()  # Prevents rotated labels from getting cut off
plt.show()
```
