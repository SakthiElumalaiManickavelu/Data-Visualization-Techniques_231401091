```python
import sqlite3
import pandas as pd

# File paths
csv_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"

# 1. Load CSV dataset
df = pd.read_csv(csv_path)

# 2. Connect to SQLite and write the DataFrame to a table named 'gpu_sales'
conn = sqlite3.connect('nvidia_sales.db')
df.to_sql('gpu_sales', conn, if_exists='replace', index=False)

# 3. Query records directly from the database using SQL
query = "SELECT * FROM gpu_sales"
df_sql = pd.read_sql(query, conn)

# Display retrieved data
print(df_sql.head())

# Close connection
conn.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Aggregate units sold by GPU Family
family_sales = df.groupby('gpu_family')['units_sold'].sum().reset_index()

# Plot bar chart matching the experiment visual style
plt.figure(figsize=(8, 5))
plt.bar(family_sales['gpu_family'], family_sales['units_sold'], color='#3399FF', edgecolor='#004C99')
plt.title('Total Units Sold by GPU Family')
plt.xlabel('GPU Family')
plt.ylabel('Units Sold')
plt.tight_layout()
plt.show()
```
