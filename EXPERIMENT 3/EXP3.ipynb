```python
import pandas as pd

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Fill missing numerical values with column means
df['units_sold'] = df['units_sold'].fillna(df['units_sold'].mean())
df['revenue_usd'] = df['revenue_usd'].fillna(df['revenue_usd'].mean())

# Fill missing categorical values with mode or a placeholder
df['bundle_addon'] = df['bundle_addon'].fillna('None')

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
import pandas as pd

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Calculate Q1, Q3, and IQR for revenue_usd
Q1 = df['revenue_usd'].quantile(0.25)
Q3 = df['revenue_usd'].quantile(0.75)
IQR = Q3 - Q1

# Define upper and lower limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['revenue_usd'] < lower) | (df['revenue_usd'] > upper)]
print("Outliers identified based on Revenue:")
print(outliers[['sale_id', 'gpu_model', 'revenue_usd']].head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Normalize numerical column 'revenue_usd'
scaler = MinMaxScaler()
df[['revenue_usd_normalized']] = scaler.fit_transform(df[['revenue_usd']])

# Display original and normalized values
print(df[['revenue_usd', 'revenue_usd_normalized']].head())
```
