```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Aggregate total units sold by GPU Model
model_units = df.groupby('gpu_model')['units_sold'].sum().reset_index().sort_values(by='units_sold', ascending=False)

# Plot bar chart
plt.figure(figsize=(10, 5))
plt.bar(model_units['gpu_model'], model_units['units_sold'], color='#3399FF', edgecolor='#004C99')
plt.title('Units Sold by GPU Model')
plt.xlabel('GPU Model')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
