```python
import pandas as pd

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Filter 1: Transactions with units sold greater than 30
high_volume = df[df["units_sold"] > 30]
print("High Volume Sales (>30 units):")
print(high_volume[["sale_id", "gpu_model", "units_sold"]].head())

# Filter 2: Transactions for Data Center AI family
datacenter_sales = df[df["gpu_family"] == "Data Center AI"]
print("\nData Center AI Sales:")
print(datacenter_sales[["sale_id", "gpu_model", "revenue_usd"]].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Sort dataset by revenue_usd in descending order
sorted_df = df.sort_values(by="revenue_usd", ascending=False)
print("Top Revenue Generating Transactions:")
print(sorted_df[["gpu_model", "revenue_usd"]].head())

# Visualizing top 10 highest revenue sales
top_10 = sorted_df.head(10)
plt.figure(figsize=(10, 5))
plt.bar(top_10["gpu_model"], top_10["revenue_usd"], color="#3399FF", edgecolor="#004C99")
plt.title("Top 10 GPU Sales Transactions by Revenue")
plt.xlabel("GPU Model")
plt.ylabel("Revenue (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Specific statistical measures for revenue_usd
print("Mean Revenue =", df["revenue_usd"].mean())
print("Median Revenue =", df["revenue_usd"].median())
print("Mode Revenue =", df["revenue_usd"].mode()[0])
print("Standard Deviation =", df["revenue_usd"].std())

print("\n--- Summary Statistics ---")
print(df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Group by GPU Family and calculate average revenue
family_avg = df.groupby("gpu_family")["revenue_usd"].mean().reset_index()
print(family_avg)

# Plot Department/Family-wise average revenue
plt.figure(figsize=(8, 5))
plt.bar(family_avg["gpu_family"], family_avg["revenue_usd"], color="#3399FF", edgecolor="#004C99")
plt.title("Average Revenue by GPU Family")
plt.xlabel("GPU Family")
plt.ylabel("Average Revenue (USD)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r"C:\Users\Sakthi\Data Visualization Techniques\nvidia_gpu_sales_synthetic_2026.csv"
df = pd.read_csv(file_path)

# Compute correlation matrix between units sold and revenue
corr = df[["units_sold", "revenue_usd"]].corr()
print("Correlation Matrix:")
print(corr)

# Scatter plot: Units Sold vs Revenue
plt.figure(figsize=(8, 5))
plt.scatter(df["units_sold"], df["revenue_usd"], color="#3399FF", alpha=0.6)
plt.title("Units Sold vs Revenue Relationship")
plt.xlabel("Units Sold")
plt.ylabel("Revenue (USD)")
plt.grid(False)
plt.tight_layout()
plt.show()
```
